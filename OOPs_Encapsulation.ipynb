```python
class BankAccount:
    def __init__(self):
        self.__balance=5000
    def withdraw(self,amount):
        
        if amount<=0:
            return " Invalid"
        if amount>self.__balance:
            return f"Insufficient amount. Balance is {self.__balance}"
           
        else:
            self.__balance= self.__balance-amount
            return f'Transaction successfull. Remaining balance is {self.__balance}'
       
    def get_balance(self):
        return f"Remaining balance is {self.__balance}"
    
```

```python
add=BankAccount()
add.withdraw(-1)
```

```python
add.get_balance()
```

```python
Q7. Employee Salary Validation
Employee class banao:
- Private attribute __salary.
- set_salary(amount) method salary set kare.
- Salary 0 se zyada ho tabhi set ho; warna "Invalid salary" print ho.
- get_salary() salary return kare.
Test: Salary 30000 aur phir -5000 set karke dekho.
```

```python
class Employee:
    def __init__(self):
       pass
    def set_salary(self,amount):
        
        if amount<0:
            return "invalid"
        else:
            self.__salary=amount
            return f"salary set {self.__salary}"
        
    def get_salary(self):
        return self.__salary
        
```

```python
sal=Employee()
```

```python
sal.set_salary(30000)
```

```python
sal.get_salary()
```

```python
Q10. Password Change System
User class banao:
- Private attribute __password.
- change_password(old_password, new_password) method ho.
- Purana password match hone par hi password change ho.
- Naya password kam se kam 8 characters ka ho.
- check_password(password) se password verify karo.
Test: Sahi aur galat old password, aur chhota new password.
```

```python
class User:
    __password='Python@123'
    
    def change_password(self,old_password,new_password):
        if  old_password==self.__password:
            if len(new_password)>=8:
              self.__password=new_password
              return " New password set"
            else:
              return " New password contains minium 8 character"
        else:
            return " Enter right password"
    def check_password(self,password):
        if password==self.__password:
            return 'Correct password'
        else:
            return 'incorrect password'
       
        
```

```python
user=User()
user.check_password('Python@123')
```

```python
user.change_password('Python@123','hon@123')
```

```python
user.check_password('ython@123')
```

```python
Q11. ATM Transaction System
ATM class banao:
- Private attributes __balance aur __pin.
- check_pin(pin) se PIN verify karo.
- withdraw(pin, amount) sirf correct PIN aur sufficient balance par chale.
- deposit(amount) sirf positive amount accept kare.
- get_balance(pin) correct PIN par hi balance return kare.
Test: Wrong PIN, insufficient balance aur successful transaction.
```

```python
class ATM:
    __balance=5000
    __pin=5645
    def withdraw(self,amount,pin):
        if 0>=amount:
            return 'Amount must be greater than 0'
        
        elif pin==self.__pin:
            if amount<=self.__balance:
                self.__balance=self.__balance-amount
                return 'Transaction successful'
            else:
                return 'Insufficient amount'
        else:
            return 'wrong pin'
    def deposite(self,amount):
         if 0>=amount:
            return 'Amount must be greater than 0'
         else:
             self.__balance=self.__balance+amount
             return f'Current balance {self.__balance}'
    def check_pin(self,pin):
        if pin==self.__pin:
            return " Pin varify"
        else:
            return " Wrong pin"
    def get_balance(self,pin):
        if pin==self.__pin:
           return self.__balance
        else:
            return 'wrong pin'
        
            
            
```

```python
atm=ATM()
```

```python
atm.withdraw(4000,5645)
```

```python
atm.get_balance(5645)
```

```python
atm.deposite(500)
```
