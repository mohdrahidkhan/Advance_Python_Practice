# Inheritance

### Basic

```python
Problem 1: Single Inheritance
Concept: Parent class aur Child class.

Task:
1. Animal naam ki parent class banao.
2. Isme eat() method banao, jo "Animal is eating" print kare.
3. Dog naam ki child class banao, jo Animal ko inherit kare.
4. Dog class mein bark() method banao, jo "Dog is barking" print kare.
5. Dog class ka object banakar dono methods call karo.
```

```python
class Animal:
    def eat(self):
        print('Animal is eating')
class Dog(Animal):
    def bark(self):
        print('Dog is barking')
        
```

```python
dog=Dog()
#animal=Animal()
```

```python
dog.bark()
dog.eat()
```

```python
Problem 2: Inherited Attributes
Concept: Parent class ke attributes ko inherit karna.

Task:
1. Person naam ki class banao.
2. __init__() mein name aur age initialize karo.
3. Student naam ki child class banao.
4. Student ka name, age aur course store karo.
5. Student ki details display karne ke liye display() method banao.
Condition: Parent class ke name aur age ko child class mein dobara initialize karne ke bajaye super() ka use karo.
```

```python
class Person:
    def __init__(self,name,age):
        self.name=name
        self.age=age
class Student(Person):
    def __init__(self,name,age,course):
        super().__init__(name,age)
        self.course=course
    def display(self):
        print(f'Name:{self.name}')
        print(f'Age:{self.age}')
        print(f'Course:{self.course}')
       

```

```python
std=Student('Rahid',33,'B.tech')
```

```python
std2=Student('Khan',55,'b.ed')
```

```python
std.display()
std2.display()
```

```python
Method Inheritance
Concept: Inherited methods aur child class ke apne methods.

Task:
1. Vehicle naam ki parent class banao.
2. Isme start() aur stop() methods banao.
3. Car naam ki child class banao.
4. Car mein drive() method banao.
5. Car ka object banakar teeno methods call karo.
```

```python
class Vehicle:
    def start(self):
        print('Vehicle is start')
    def stop(self):
        print('Vehicle is stop')
class Car(Vehicle):
    def drive(self):
        print('Car is driving')
```

```python
car=Car()
```

```python
car.start()
car.stop()
car.drive()
```

```python
Problem 4: Multilevel Inheritance
Concept: Ek class se doosri aur doosri se teesri class inherit karna.

Task:

Teen classes banao:
- Person: name attribute aur show_name() method.
- Employee: salary attribute aur show_salary() method.
- Manager: department attribute aur show_department() method.
```

```python
class Person:
    def __init__(self,name):
        self.name=name
    def show_name(self):
         print( f'Name:{self.name}')
class Employee(Person):
    def __init__(self,name,salary):
        super().__init__(name)
        self.salary=salary
    def show_salary(self):
        print(f'Salary:{self.salary}')
class Manager(Employee):
    def __init__(self,name,salary,department):
        super().__init__(name,salary)
        self.department=department
    def show_department(self):
        print( f'Department:{self.department}')
```

```python
man=Manager('Rahid',30000,'IT')
```

```python
man.show_name()
man.show_salary()
man.show_department()
```

```python
man.show_name()
```

```python
man.show_salary()
```

### Medium

```python
1. Employee Salary System

Concepts: Single Inheritance, super(), Method Overriding

Ek parent class Employee banao.

Requirements:
- Employee mein name aur basic_salary attributes hon.
- calculate_salary() method basic salary return kare.
- Developer class ko Employee se inherit karo.
- Developer ko bonus attribute do.
- Developer mein calculate_salary() override karo, jo basic salary + bonus return kare.
- Dono classes ke objects banakar salary print karo.
```

```python
class Employee:
    def __init__(self,name,basic_salary):
        self.name=name
        self.basic_salary=basic_salary
    def calculate_salary(self):
        return self.basic_salary
class Developer(Employee):
    def __init__(self,name,basic_salary,bonus):
        super().__init__(name,basic_salary)
        self.bonus=bonus
    def calculate_salary(self):
        return self.basic_salary+self.bonus
    
```

```python
emp = Employee("Rahul", 30000)
dev = Developer("Aman", 40000, 10000)

print(emp.calculate_salary())
print(dev.calculate_salary())
```

```python
2. Bank Account System

Concepts: Inheritance, Method Overriding, Encapsulation basics

Ek parent class BankAccount banao.

Requirements:
- Attributes: account_holder, balance
- Methods: deposit(amount), withdraw(amount), display_balance()
- SavingsAccount class inherit karo.
- Savings account mein minimum_balance add karo.
- Withdrawal tabhi allow ho jab withdrawal ke baad balance minimum balance se kam na ho.
- Invalid withdrawal par appropriate message print karo.
```

```python
class Bank_Account:
    def __init__(self):
        self.balance=10000
        
    def deposite(self,amount):
        if amount<=0:
            return "Please enter a valid amount"
        else:
            self.balance+=amount
            return self.balance
    def withdraw(self,amount):
        if amount<=0:
            return "Please enter a valid amount"
        else:
            self.balance-=amount
            return f'Transaction successful. Remaining balance {self.balance}'
    def display_balance(self):
        return self.balance
class Savings_Account(Bank_Account):
    
    def __init__(self):
        super().__init__()
        self.minimum_balance=1000
         
    def deposite(self,amount):
        if amount<=0:
            return "Please enter a valid amount"
    def withdraw(self,amount):
        if amount<=0:
            return "Please enter a valid amount"
        elif (self.balance-amount)<self.minimum_balance :
            print(f'please enter less than{self.balance-self.minimum_balance}')
        else:
            self.balance-=amount
            return f'Transaction successful. Remaining balance {self.balance}'
```
